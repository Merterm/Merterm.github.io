## Option A: HuggingFace Inference API

1. Create account at https://huggingface.co
2. Go to Settings → Access Tokens → Create new token (Read access)
3. Paste token below

```python
from huggingface_hub import InferenceClient

HF_TOKEN = "YOUR_HUGGINGFACE_TOKEN"  # <-- PASTE YOUR TOKEN

client = InferenceClient(api_key=HF_TOKEN)

def query_hf(prompt, model="deepseek-ai/DeepSeek-V3.2"):
    completion = client.chat.completions.create(
                model=model,
                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
            )

    return completion.choices[0].message.content
```

```python
# Test HuggingFace
response = query_hf("What is 2+2?")
print(response)
```

## Option B: Google Gemini API

1. Go to https://aistudio.google.com/app/apikey
2. Click "Create API Key"
3. Paste key below

```python
from google import genai
from google.genai import types

GEMINI_KEY = "YOUR_GEMINI_API_KEY"  # <-- PASTE YOUR KEY

# Initialize the client with the new SDK
client = genai.Client(api_key=GEMINI_KEY)

def query_gemini(prompt, model="gemini-2.5-flash"):
    try:
        # Create a configuration for safety settings (BLOCK_NONE for experimentation)
        # Note: 'safety_settings' is now a list of SafetySetting objects inside GenerateContentConfig
        config = types.GenerateContentConfig(
            safety_settings=[
                types.SafetySetting(
                    category="HARM_CATEGORY_HARASSMENT",
                    threshold="BLOCK_NONE"
                ),
                types.SafetySetting(
                    category="HARM_CATEGORY_HATE_SPEECH",
                    threshold="BLOCK_NONE"
                ),
                types.SafetySetting(
                    category="HARM_CATEGORY_SEXUALLY_EXPLICIT",
                    threshold="BLOCK_NONE"
                ),
                types.SafetySetting(
                    category="HARM_CATEGORY_DANGEROUS_CONTENT",
                    threshold="BLOCK_NONE"
                )
            ]
        )

        # The new call structure uses client.models.generate_content
        response = client.models.generate_content(
            model=model,
            contents=prompt,
            config=config
        )
        
        # In the new SDK, response.text is still the way to access the output
        return response.text

    except Exception as e:
        return f"Error: {e}"
```

```python
# Test Gemini
response = query_gemini("What is 2+2?")
print(response)
```
